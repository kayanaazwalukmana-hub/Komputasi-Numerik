# Komputasi Numerik: Mencari Solusi SPL Tiga Variabel

Notebook ini menghitung nilai x, y, dan z dari tiga persamaan linear
memakai **eliminasi Gauss** yang dilengkapi pemilihan pivot terbesar.

Persamaan yang diselesaikan:

- 3x + 2y − z = 7
- x − y + 2z = 5
- 2x + 3y + z = 12

Run tiap cell berurutan dari atas.

### Langkah 1: Siapkan data persamaan

```python
koef = [[3, 2, -1],
        [1, -1, 2],
        [2, 3, 1]]

ruas_kanan = [7, 5, 12]

jml_var = len(ruas_kanan)
print("Koefisien :", koef)
print("Ruas kanan:", ruas_kanan)
```

### Langkah 2: Bentuk matriks gabungan

```python
def tampilkan(tabel, keterangan):
    print(keterangan)
    for brs in tabel:
        print(" | ".join(f"{angka:9.4f}" for angka in brs))
    print()

tabel = []
for i in range(jml_var):
    tabel.append(koef[i] + [ruas_kanan[i]])

tampilkan(tabel, "Matriks gabungan (awal):")
```

### Langkah 3: Eliminasi ke bentuk segitiga atas

```python
for kol in range(jml_var - 1):
    idx_terbesar = kol
    for r in range(kol + 1, jml_var):
        if abs(tabel[r][kol]) > abs(tabel[idx_terbesar][kol]):
            idx_terbesar = r

    if idx_terbesar != kol:
        tabel[kol], tabel[idx_terbesar] = tabel[idx_terbesar], tabel[kol]
        print(f"-> baris {kol+1} ditukar dengan baris {idx_terbesar+1}")

    if abs(tabel[kol][kol]) < 1e-12:
        raise ValueError("Tidak ada solusi tunggal (pivot bernilai nol)")

    for r in range(kol + 1, jml_var):
        pengali = tabel[r][kol] / tabel[kol][kol]
        for c in range(kol, jml_var + 1):
            tabel[r][c] = tabel[r][c] - pengali * tabel[kol][c]

    tampilkan(tabel, f"Hasil setelah kolom {kol+1} dieliminasi:")
```

### Langkah 4: Hitung variabel dari bawah ke atas

```python
if abs(tabel[-1][-2]) < 1e-12:
    raise ValueError("Tidak ada solusi tunggal")

solusi = [0.0] * jml_var
for r in reversed(range(jml_var)):
    total = 0.0
    for c in range(r + 1, jml_var):
        total += tabel[r][c] * solusi[c]
    solusi[r] = (tabel[r][-1] - total) / tabel[r][r]

for nama, nilai in zip("xyz", solusi):
    print(f"{nama} = {nilai:.4f}")
```

### Langkah 5: Cek hasil ke persamaan awal

```python
for r in range(jml_var):
    hitung = sum(koef[r][c] * solusi[c] for c in range(jml_var))
    status = "sesuai" if abs(hitung - ruas_kanan[r]) < 1e-9 else "TIDAK sesuai"
    print(f"Persamaan {r+1}: {hitung:.4f} vs {ruas_kanan[r]} -> {status}")
```
